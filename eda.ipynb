```python
import pandas

data = pandas.read_csv("datasets/supermarket_data.csv")
data
```

```python
# use the Describe function
data.describe()
```

```python
#we have noticed that the column itemprice column, contains three empty slots
# we need to fill this empty slots with a value and in this case we shall fill with the mean of the column
data["itemprice"] = data["itemprice"].fillna(55)

#describe data again
data.describe()
```

```python
# get the number of entries in a given column
data.groupby("gender").size()
```

```python
# combining multiple columns 
data.groupby(["city", "payment"]).size()
```

```python
#import the modules
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# below is a count plot showing the distribution of males to females
sns.countplot(x="gender", data=data,palette="dark")
plt.title("Gender Distribution Against Total count")
```

```python
# create a count plot of product
sns.countplot(x = "product", data=data,palette="dark")
plt.xticks(rotation = 45)
plt.title("Distribution of products")
```

```python
#below is a countplot that has two columns
sns.countplot(x="gender", hue = "city", palette="dark", data =data)
```

```python
# below is a pie plot
data["payment"].value_counts().plot.pie(autopct='%1.1f%%')
```
